```python
test_payments = [
    {"order_id": 301, "amount": ""},
    {"order_id": 302, "amount": " 300 "},
    {"order_id": 303, "amount": "0"},
]
```

```python
def clean_orders(orders):
    valid_orders = []
    rejected_orders = []
    for i in orders:
        try:
            new_i = i.copy()
            new_i.update({'amount' : int(new_i['amount'])})
            valid_orders.append(new_i)
        except ValueError as e:
            new_i.update({'reason' : 'The amount is not a valid one'})
            rejected_orders.append(new_i)
    return valid_orders,rejected_orders
```

```python
valid_result, rejected_result = clean_orders(test_payments)
```

```python
valid_result
```

```python
rejected_result
```
