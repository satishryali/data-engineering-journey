```python
from faker import Faker
import csv
from faker_commerce import Provider
import random
```

```python
fake = Faker("en_IN")
```

```python
fake.add_provider(Provider)
```

```python
with open('product_data.csv','w') as file:
    writer = csv.writer(file)
    writer.writerow(['product_id','product_name','category','selling_price','created_at'])
    for i in range(1000):
        writer.writerow([i + 1,fake.ecommerce_name(),fake.ecommerce_category(),fake.ecommerce_price(random.randint(5000,50000)),fake.date_time()])
        
```
