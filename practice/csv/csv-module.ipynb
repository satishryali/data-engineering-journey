```python
with open('product_data.csv','w') as file:
    writer = csv.writer(file)
    writer.writerow(['id','name','price'])
    writer.writerow([1,'towel',20])
    writer.writerows([[2,'shirt',2000],[3,'trowser',600],[4,'whiteshirt',205]])
```

```python
field_name = ['id','name','price']
with open('product_data_dict.csv','w') as file:
    writer = csv.DictWriter(file,fieldnames=field_name)
    writer.writeheader()
    writer.writerow({'id' : 6,'name' : 'Table','price' : 600})
    
```
