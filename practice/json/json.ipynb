```python
import json 

text = """ {
"customer_id" : 101,
"name" : "Anu",
"orders" : [{"order_id" : 1, "amount" : 250},
{"order_id" : 2, "amount" : 400}]}"""

data = json.loads(text)
```

```python
print(type(data),type(data["orders"]))
print(data["name"])
total_amount = 0
for i in data["orders"]:
    print(i["order_id"], i["amount"])
    total_amount += i["amount"]
print(total_amount)
print(type(text))
```

```python
with open("customer.json",'w',newline="") as file:
    json.dump(data,file,indent=2)
```

```python
with open('customer.json','r',encoding="utf-8",newline="") as file:
    loaded_data = json.load(file)
    print(loaded_data)
```

```python
total_amount = 0
print("name : " , loaded_data["name"])
for i in loaded_data["orders"]:
    total_amount += i["amount"]

print("Total_Amount : ", total_amount)

```

```python
import csv
import json 

orders = []
Record_count = 0
with open('orders.csv','r',encoding='utf-8') as file:
    reader = csv.DictReader(file)
    for i in reader:
        orders.append({
            "order_id" : int(i["order_id"]),
            "customer_name" : i["customer_name"],
            "amount" : float(i["amount"]),
            "status" : i["status"],
        })
with open('orders.json','w',encoding='utf-8') as file:
    json.dump(orders,file,allow_nan=False,indent=2,ensure_ascii=False)

with open('orders.json','r',encoding='utf-8') as file:
    loaded_data = json.load(file)
    for i in loaded_data:
        Record_count += 1
print(Record_count)
```
