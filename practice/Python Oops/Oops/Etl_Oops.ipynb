```python
import csv

def load(data,filename1,filename2,filename3):
    design = data
    orders = []
    cancelled_orders = []
    cancelled_count = 0
    cancelled_total = 0
    completed_count = 0
    completed_total = 0
    with open(data,'r') as file:
        reader = csv.DictReader(file)
        for i in reader:
            orders.append(i)
        fieldname = ['order_id','customer_id','quantity','unit_price','total_amount']
        orsum_field = ['status','order_count','total_amount']
        with (
            open(f'{filename1}','w',newline="") as complete_file,
            open(f'{filename2}','w',newline="") as cancel_file,
            open(f'{filename3}','w',newline="") as order_summary,
            ):
            writer = csv.DictWriter(complete_file,fieldnames=fieldname)
            ca_writer = csv.DictWriter(cancel_file,fieldnames=fieldname)
            orsum_writer = csv.DictWriter(order_summary,fieldnames=orsum_field)
            writer.writeheader()
            ca_writer.writeheader()
            orsum_writer.writeheader()
            for i in orders:
                if i['status'] == 'completed':
                    writer.writerow({'order_id' : i['order_id'],
                                    'customer_id' : i['customer_id'],
                                    'quantity' : i['quantity'],
                                    'unit_price' : i['unit_price'],
                                    'total_amount' : int(i['quantity']) * int(i['unit_price'])})
                elif i['status'] == 'cancelled':
                    ca_writer.writerow({'order_id' : i['order_id'],
                                    'customer_id' : i['customer_id'],
                                    'quantity' : i['quantity'],
                                    'unit_price' : i['unit_price'],
                                    'total_amount' : int(i['quantity']) * int(i['unit_price'])})
            for i in orders:
                if i['status'] == 'completed':
                    completed_count += 1
                    completed_total += int(i['quantity']) * int(i['unit_price'])
                elif i['status'] == 'cancelled': 
                    cancelled_count += 1
                    cancelled_total += int(i['quantity']) * int(i['unit_price'])
                
            orsum_writer.writerow({'status' : 'completed',
                                'order_count' : completed_count,
                                'total_amount' : completed_total})
            orsum_writer.writerow({'status' : 'cancelled',
                                                    'order_count' : cancelled_count,
                                                    'total_amount' : cancelled_total})

load('orders.csv','completed_orders.csv','cancelled_orders.csv','orders_summary.csv')

load('orders_day2.csv','completed_orders_day2.csv','cancelled_orders_day2.csv','orders_summary_day2.csv')  

load('orders_day3.csv','completed_orders_day3.csv','cancelled_orders_day3.csv','orders_summary_day3.csv')  

                
            
        
```
