```python
from dataclasses import dataclass

@dataclass
class Order:
    order_id : int
    status : str
    quantity : int

order = Order(
    order_id = "1001",
    status = "completed",
    quantity = 3
)

print(order)
print(order.quantity)
```

```python
class Customer:
    def __init__(self,name):
        self.name = name

    def __str__(self):
        return self.name

    def __repr__(self):
        return f"Customer(name = {self.name})"
obj = [Customer('satish'),Customer('ravi')]
print(str(obj))
```

```python
class Customeronlystr:
    def __init__(self,name):
        self.name = name

    def __repr__(self):
        return f"Customer({self.name})"

    def __str__(self):
        return self.name

c = Customeronlystr('satish')
```

```python
class Quantity:
    def __init__(self,quantity):
        self.quantity = quantity

    def __eq__(self,other):
        return self.quantity == other.quantity

order1 = Quantity('satish')
order2 = Quantity('satish')

print(order1.__eq__(order2))
```

```python
class Quantity1:
    def __init__(self,quantity):
        self.quantity = quantity

order1 = Quantity1('satish')
order2 = Quantity1('satish')

order1 == order2
```

```python
class Order:
    def __init__(self,quantity):
        self.quantity = quantity

    def __lt__(self,other):
        return self.quantity > other.quantity

first = Order(10)
second = Order(20)

print(first < second)
```

```python
class Order:
    def __init__(self,quantity):
        self.quantity = quantity

    def __gt__(self,other):
        return self.quantity > other.quantity

first = Order(10)
second = Order(20)

second.__gt__(first)
```

```python
class Batch:
    def __len__(self):
        return -1

batch = Batch()
print(len(batch))
```

```python
class Batch:
    def __getitem__(self,position):
        rows = ["order-A","order-B"]
        return rows[position]

batch = Batch()
print(batch[1])
```
