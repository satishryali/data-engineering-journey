```python
import time

def fetch_users():
    print("1.Started fetching users...")
    time.sleep(2)
    print("2. Done fetching users..")
    return ["Alice","Bob"]

def fetch_orders():
    print("3. Started fetching orders...")
    time.sleep(2)
    print("4. Done fetching orders!..")
    return [101,102]


start = time.time()

users = fetch_users()
orders = fetch_orders()

total_time = time.time() - start

print(f"\n finished everything in {total_time:.2f} seconds")
```

```python
import time
from concurrent.futures import ThreadPoolExecutor

start = time.time()

with ThreadPoolExecutor(max_workers=2) as executor:
    future_users = executor.submit(fetch_users)
    future_orders = executor.submit(fetch_orders)

    users = future_users.result()
    orders = future_orders.result()

total_time = time.time() - start

print(f"users : {users}")
print(f"Orders : {orders}")
print(f"Finished everything in {total_time:.2f} seconds")
```

```python
import time
import threading

def slow_work():
    print(f"inside function before sleep running on -> {threading.current_thread().name}")
    time.sleep(3)
    print("I am threading , running in the background using a thread")
    print(f"inside function running after sleep on -> {threading.current_thread().name}")

print("I am not a thread, running directly.")
print(f"outisde the function before start -> {threading.current_thread().name}")
t = threading.Thread(target = slow_work)
t.start()
print(f"outside the fucntion after the start -> {threading.current_thread().name}")
```

```python
import threading
import time
import csv
list = []
field_name = ['product_id','product_name','category','discounted_price','actual_price','discount_percentage','rating','rating_count','about_product','user_id','user_name','review_id','review_title','review_content','img_link','product_link']
def loading_data():
    print(f"Loading the sales_data : {threading.current_thread().name}")
    with open("amazon.csv",'r',encoding="utf-8",newline="") as file:
        reader = csv.DictReader(file)
        for i in reader:
            list.append(i)
        print(f"reading completed : {threading.current_thread().name}")

    print(f"Writing the data to the amazon_v1.csv : {threading.current_thread().name}")
    with open("amazon_v1.csv",'w',newline="",encoding="utf-8") as file:
        writer = csv.DictWriter(file,fieldnames=field_name)
        writer.writeheader()
        writer.writerows(list)

        print(f"writing to the file is completed : {threading.current_thread().name}")

print(f"Before the start : {threading.current_thread().name}")
t = threading.Thread(target=loading_data)
t.start()
t.join()
print(f"Before reading after start : {threading.current_thread().name}")
with open('amazon_v1.csv','r',encoding='utf-8',newline="") as file:
    reader = csv.reader(file)
    for i in reader:
        print(i)
print(f"After reading after start : {threading.current_thread().name}")

print(f"After the start : {threading.current_thread().name}")
```
