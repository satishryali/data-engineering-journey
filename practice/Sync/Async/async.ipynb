```python
import threading
import time
import csv
list = []
field_name = ['product_id','product_name','category','discounted_price','actual_price','discount_percentage','rating','rating_count','about_product','user_id','user_name','review_id','review_title','review_content','img_link','product_link']
def loading_data():
    print(f"Loading the sales_data : {threading.current_thread().name}")
    with open("amazon.csv",'r',encoding="utf-8",newline="") as file:
        reader = csv.DictReader(file)
        for i in reader:
            list.append(i)
        print(f"reading completed : {threading.current_thread().name}")

    print(f"Writing the data to the amazon_v1.csv : {threading.current_thread().name}")
    with open("amazon_v1.csv",'w',newline="",encoding="utf-8") as file:
        writer = csv.DictWriter(file,fieldnames=field_name)
        writer.writeheader()
        writer.writerows(list)

        print(f"writing to the file is completed : {threading.current_thread().name}")

print(f"Before the start : {threading.current_thread().name}")
t = threading.Thread(target=loading_data)
t.start()
t.join()
print(f"Before reading after start : {threading.current_thread().name}")
with open('amazon_v1.csv','r',encoding='utf-8',newline="") as file:
    reader = csv.reader(file)
    for i in reader:
        print(i)
print(f"After reading after start : {threading.current_thread().name}")

print(f"After the start : {threading.current_thread().name}")
```

```python
import time
from concurrent.futures import ThreadPoolExecutor

def fetch_orders(shop):
    print(f"{shop} : started")
    time.sleep(2)
    print(f"{shop} : finished")
    return f"Orders from {shop}"

start = time.perf_counter()

print(start)

with ThreadPoolExecutor(max_workers=2) as executor:
    job_a = executor.submit(fetch_orders,"Shop A")
    job_b = executor.submit(fetch_orders,"Shop B")
    job_c = executor.submit(fetch_orders,"Shop C")

    orders_a = job_a.result()
    orders_b = job_b.result()
    orders_c = job_c.result()

print(orders_a)
print(orders_b)
print(orders_c)
print(f"Total time : {time.perf_counter() - start:.2f} seconds")
```

```python
import json
from urllib.parse import urlencode
from urllib.request import urlopen
from concurrent.futures import ThreadPoolExecutor

def get_temperature(city,latitude,longitude):
    parameters = urlencode({
        "latitude" : latitude,
        "longitude" : longitude,
        "current" : "temperature_2m",
    })
    url = f"https://api.open-meteo.com/v1/forecast?{parameters}"

    print(f"Requesting {city}...")
    with urlopen(url,timeout=10) as response:
        data = json.load(response)
    return city,data["current"]["temperature_2m"]

with ThreadPoolExecutor(max_workers=2) as executor:
    hyderabad = executor.submit(get_temperature,"Hyderabad",17.285,78.4867)
    mumbai = executor.submit(get_temperature,"Mumbai",19.076,72.8777)
    bengaluru = executor.submit(get_temperature,"Bengaluru",12.9716,77.5946)

    for job in (hyderabad,mumbai,bengaluru):
        city,temperature = job.result()
        print(f"{city} : {temperature} c")
```

```python
import sys
import time
from concurrent.futures import ThreadPoolExecutor

def calculate(limit):
    total = 0
    for number in range(limit):
        total += number * number

limit = 5_000_000

print("Gil Enabled : ",sys._is_gil_enabled())

start = time.perf_counter()
first = calculate(limit)
second = calculate(limit)
print("Sequential time: ",round(time.perf_counter() - start, 2))

start = time.perf_counter()
with ThreadPoolExecutor(max_workers=2) as executor:
    first_job = executor.submit(calculate,limit)
    second_job = executor.submit(calculate,limit)
    threaded_first = first_job.result()
    threaded_second = second_job.result()

print("Threaded time : ", round(time.perf_counter() - start,2))

print("Same results : ", (first,second) == (threaded_first,threaded_second))
```

```python
from concurrent.futures import ThreadPoolExecutor
import time
import sys

def calculate(limit):
    total = 0
    for i in range(limit):
        total += i * i
    return total
limit = 5_000_000

#Sequential timer
start = time.perf_counter()
first = calculate(limit)
second = calculate(limit)
print("Sequential result : ", first)
print("Sequential_result : ", second)
end = time.perf_counter()
Sequential_time = round(end - start,2)
print(Sequential_time)

print("GiL ", sys._is_gil_enabled())
#multithreading
start = time.perf_counter()
with ThreadPoolExecutor(max_workers=2) as executor:
    work1 = executor.submit(calculate,limit)
    work2 = executor.submit(calculate,limit)
    print("Threading result  : ", work1.result())
    print("Threading result  : ", work2.result())
    thread_time = round(time.perf_counter() - start,2)
print("threading timer : ", thread_time)

```

```python
import asyncio
import time

async def fetch(shop):
    print(f"{shop} : started")
    await asyncio.sleep(2)
    print(f"{shop}: finished")
    return f"Orders from shop"

async def main():
    start = time.perf_counter()
    first = await fetch("Shop A")
    second = await fetch("Shop B")
    print("One after another : ", first,second)
    print(f"Time : {time.perf_counter() - start:.2f} seconds")

    start = time.perf_counter()
    first,second = await asyncio.gather(
        fetch("Shop A"),
        fetch("Shop B"),
    )
    print("Together:",first,second)
    print(f"Time: {time.perf_counter() - start:.2f} seconds")

await main()
```

```python
import asyncio
import time

async def fetch(shop):
    print(f"{shop} : request started")
    await asyncio.sleep(10)
    print(f"{shop}: response ready")
    await asyncio.sleep(10)
    return f"Orders from {shop}"

start = time.perf_counter()
shop_a = asyncio.create_task(fetch("Shop A"))
shop_b = asyncio.create_task(fetch("Shop B"))

print(await shop_a)
print(await shop_b)
print(f"Time : {time.perf_counter() - start:.2f} seconds")
```
